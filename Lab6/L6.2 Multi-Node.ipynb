# Poly Regression with TensorFlow (Keras)
This notebook demonstrates a simple linear regression example using TensorFlow and Keras. We will:
- Generate synthetic data
- Build and train a deep learning model
- Visualize the results and training loss.

## Importing Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD, Adam
```

## Generating Synthetic Data
We generate data based on the equation $y = x^2 + \text{noise}$. This helps us simulate real-world data with some randomness.

```python
np.random.seed(42)
X = 8 * np.random.rand(100, 1) - 4  # Input feature
y = X**2 + np.random.randn(100, 1)  # True line: y = x^2 + noise

plt.scatter(X, y, label="Data")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

## Building the Linear Model
We use Keras to create a simple linear regression model with a single dense layer.

```python
model = Sequential([
    Dense(4, input_dim=1, activation='relu'),  # Hidden layer with ReLU activation
    Dense(1)                                   # Output layer
])
```

```python
# Show the detail of model
model.summary()
```

## Checking Initial Weights and Bias
Before training, we inspect the initial weights and bias of the model.

```python
# Display weights and biases for each layer
for i, layer in enumerate(model.layers):
    weights, biases = layer.get_weights()  # Get weights and biases
    print(f"Layer {i + 1} - {layer.name}")
    print("Weights:")
    print(weights)
    print("Biases:")
    print(biases)
    print("-" * 50)
```

## Compiling the Model
We compile the model with the Stochastic Gradient Descent (SGD) optimizer and Mean Squared Error (MSE) loss.

```python
model.compile(optimizer=SGD(learning_rate=0.01), loss='mean_squared_error')
```

## Training the Model
We train the model for 500 epochs and record the training loss.

```python
history = model.fit(X, y, epochs=500, verbose=1)
```

## Trained Weights and Bias
After training, we display the updated weights and bias.

```python
# Display weights and biases for each layer
for i, layer in enumerate(model.layers):
    weights, biases = layer.get_weights()  # Get weights and biases
    print(f"Layer {i + 1} - {layer.name}")
    print("Weights:")
    print(weights)
    print("Biases:")
    print(biases)
    print("-" * 50)
```

## Making Predictions
Using the trained model, we predict outputs for the input data.

```python
y_pred = model.predict(X)
```

## Visualizing the Results
We plot the original data points and the model's predictions.

```python
plt.scatter(X, y, label="Data")
plt.scatter(X, y_pred, color="red", label="Prediction")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

## Plotting the Training Loss
We visualize how the loss decreases over epochs.

```python
plt.plot(history.history['loss'])
plt.title('Model Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.show()
```

# Try with the More Complex Polynomial line

## Generating Synthetic Data
We generate data based on the equation $y = x^3 - 2x^2 + x - 10 +\text{noise}$. This helps us simulate real-world data with some randomness.

```python
np.random.seed(42)
X = 6 * np.random.rand(100, 1) - 2  # Input feature
y = X**3 - 2*X**2 + X - 10 + np.random.randn(100, 1) 

plt.scatter(X, y, label="Data")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

## Build Model and Training again

```python
model = Sequential([
    Dense(4, input_dim=1, activation='relu'),  # Hidden layer with ReLU activation
    Dense(1)                                   # Output layer
])

# Show the detail of model
model.summary()
```

```python
# Display weights and biases for each layer
for i, layer in enumerate(model.layers):
    weights, biases = layer.get_weights()  # Get weights and biases
    print(f"Layer {i + 1} - {layer.name}")
    print("Weights:")
    print(weights)
    print("Biases:")
    print(biases)
    print("-" * 50)
```

```python
model.compile(optimizer=Adam(learning_rate=0.01), loss='mean_squared_error')
history = model.fit(X, y, epochs=500, verbose=1)
```

```python
y_pred = model.predict(X)
plt.scatter(X, y, label="Data")
plt.scatter(X, y_pred, color="red", label="Prediction")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```
